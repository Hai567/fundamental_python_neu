```python
a = "wiffffffiiiiii"
vowels = "ueoai"

for i in a:
    if i in vowels:
        a = a.replace(i*2, i)
    else:
        a = a.replace(i*3, i*2)
    
print(a)
```

```python
str = "abc-123-xyz456asfvc199"
eq = ""

for i in range(len(str)):
    if str[i].isalpha():
        if str[(i+1)%len(str)].isdigit():
            eq += "+"
    else:
        eq += str[i]

print(eq)

eval(eq)
```
