```python
class ShoppingCart:
    def __init__(self):
        self.items = {}
        
    def add_item(self, name, price):
        if "products" not in self.items:
            self.items["products"] = []
        self.items["products"].append({"name": name, "price": float(price)})

    def get_total(self):
        return sum(p["price"] for p in self.items.get("products", []))

    def apply_discount(self, percent):
        if not (0 <= percent <= 100):
            raise ValueError("percent must be between 0 and 100")
        factor = (100 - percent) / 100
        for p in self.items.get("products", []):
            p["price"] *= factor
        return self.get_total()

cart = ShoppingCart()
cart.add_item("Milk", 10.0)
print(cart.get_total())
print(cart.apply_discount(10))
```

```python
def count_vowels(text):
    c = 0
    for i in text:
        if i in "ueoai":
            c += 1
    return c
```

```python
def merge_inventories(inv1, inv2):
    final = inv1
    for k, v in inv2.items():
        if k.lower() in [k2.lower() for k2, _ in inv2.items()]:
            final[k.lower()] += v
        else:
            final[k.lower()] = v
    return final
            
            
print(sorted(merge_inventories({"apple": 5}, {"apPle": 10}).items()))
```

```python
class Message:
    def __init__(self, recipient, sender):
        self.recipient = recipient
        self.sender = sender
        self.msg = ""
        
    def append(self, text):
        self.msg += f"{text}\n"

    def __str__(self):
        return f"From: {self.sender}\nTo: {self.recipient}\n{self.msg}"
    
class MailBox():
    def __init__(self):
        self._msgs = []
    
    def addMsg(self, msg):
        if isinstance(msg, Message):
            self._msgs.append(msg)
            
    def getMsg(self, index):
        return self._msgs[index]
    
    def removeMsg(self, index):
        return self._msgs.pop(index)
    
    
m1 = Message("you", "me")
m1.append("Hello")
m2 = Message("you", "me")
m2.append("Hello 2")

mailbox = MailBox()
mailbox.addMsg(m1)
mailbox.addMsg(m2)
print(mailbox._msgs)

print(mailbox.getMsg(0))
print(mailbox.removeMsg(0))
print(mailbox._msgs)
```
