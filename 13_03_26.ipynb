Ex 1:

Input: Text of numbers separated by " "
Output: The list of those numbers

```python
def textToArray(text):
    return text.split(" ")

textInput = input("Enter a list of number separated by space: ")
print(textToArray(textInput))
```

Ex 2: Count how many time an element appear in an array

Input: a, list
Output: How many time "a" appears in "list"

```python
def countElementInList(a, list):
    count = 0
    for element in list:
        if element == a:
            count +=1
    return count

textInput = input("Enter a list of number separated by space: ")
a = input("Enter a number to count: ")

numberList = textToArray(textInput)
print(numberList)

print(countElementInList(a, numberList))
```
