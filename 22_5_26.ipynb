- Order:
    - OrderNumber
    - ProductItems [ProductItem1, ProductItem2]
    - TotalPayment
    - addProductItem(ProductItem)
    - updateQuantity(productCode, newQuant)
    - calculaatePaymentAmount()

- ProductItem:
    - Product
    - Quant

- Product:
    - Code
    - Name
    - Unit
    - Price

```python
class Product:
    def __init__(self, code, name, unit, price):
        self.code = code
        self.name = name
        self.unit = unit
        self.price = price
        
class ProductItem():
    def __init__(self, product, quant):
        self.product = product
        self.quant = quant
    
class Order:
    def __init__(self, ordernum):
        self.orderNum = ordernum
        self.prodItems = []
        self.totalPayment = 0

    def addProdItem(self, prodItem):
        if isinstance(prodItem, ProductItem):
            self.prodItems.append(prodItem)

    def updateQuant(self, prodCode, newQuant):
        for prod in self.prodItems:
            if prod.product.code == prodCode:
                prod.quant = newQuant
                break
        
    
    def calculatePaymentAmount(self):
        total = 0
        for prod in self.prodItems:
            total += prod.quant * prod.product.price 
        self.totalPayment = total
    
    
apple = Product("1", "Apple", "kg", 3.5)

appleItem = ProductItem(apple, 2)

myOrder = Order("001")
myOrder.addProdItem(appleItem)
print(myOrder.calculatePaymentAmount())
```
